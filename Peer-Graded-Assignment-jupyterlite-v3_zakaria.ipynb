<h3>Question 1</h3>
Is there a significant difference in median value of houses bounded by the Charles River or not? (T-test for independent samples).

* **CHAS** – Charles River dummy variable (1 if tract bounds river; 0 otherwise)
* **MEDV** – Median value of owner-occupied homes in \$1000's

1. **State your hypothesis**

    * $H_0: µ_1 = µ_2$
        * There is **NOT** a difference in the median value of houses bounded by the Charles River.

    * $H_1:$
        * There **IS** a difference in the median value of houses bounded by the Charles River.
2. Use α = 0.05
3. Perform the test Statistics.

```python
chas_0 = boston_df[boston_df['CHAS'] == 0]['MEDV']
chas_1 = boston_df[boston_df['CHAS'] == 1]['MEDV']

t_stat, p_value = ttest_ind(chas_0, chas_1, equal_var=False)

print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.8f}")
```

4. State the conclusion from the test
    * Since P-value: 0.0036 < α = 0.05, we reject the null hypothesis. There is a difference in the median value of houses by the Charles River.

Can we conclude that there is no relationship between Nitric oxide concentrations and proportion of non-retail business acres per town? (Pearson Correlation)

* **NOX** -  Nitric oxides concentration (parts per 10 million)

1. **State your hypothesis**

<h3>Question 4</h3>
What is the impact of an additional weighted distance to the five Boston employment centres on the median value of owner occupied homes? (Regression analysis)

* **DIS** – Weighted distances to five Boston employment centres.
* **MEDV** – Median value of owner-occupied homes in \$1000's

1. **State your hypothesis**

    * $H_0: µ_1 = µ_2$
        * There is **NO RELATIONASHIP** between distance to employment centers (DIS) and median home value (MEDV).

    * $H_1:$
        * There **IS A RELATIONASHIP** between distance to employment centers (DIS) and median home value (MEDV).
2. Use α = 0.05
3. Perform the test Statistics.

```python
X = boston_df['DIS']
y = boston_df['MEDV']

X = sm.add_constant(X)

model = sm.OLS(y, X).fit()
print(model.summary())
```

4. State the conclusion from the test

        *  There **IS A RELATIONASHIP** between distance to employment centers (DIS) and median home value (MEDV).
